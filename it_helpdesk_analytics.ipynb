```python
# create a function for displaying query results.

def execute_query(q):
    cursor.execute(q)

    result = cursor.fetchall()

    for row in result:
        print(row)

    print()
    
```

Industry-Oriented QUESTIONS

```python
# 14. The IT manager wants to see all Open tickets along with their priority and issue type. Retrieve the required information.
q = """select ticket_id, priority, issue_type from support_tickets where status = 'Open'"""
execute_query(q)
```

```python
# 15. The support team wants to find all employees who have raised a ticket related to Email issues.
q = """select e.employee_name, i.issue_type from employees e join support_tickets i on e.employee_id = i.employee_id where i.issue_type = 'Email'"""
execute_query(q)
```

```python
# 16. Display the employee name, department, ticket ID, issue type, priority, and status for each support ticket.
q = """select e.employee_name, e.department, s.ticket_id, s.issue_type, s.priority, s.status from employees e join support_tickets s on e.employee_id = s.employee_id"""
execute_query(q)
```

```python
# 17. Find the employees who have raised more than one support ticket.
q= """select e.employee_name, count(s.ticket_id) as ticket_count from employees e join support_tickets s on e.employee_id = s.employee_id group by e.employee_name having ticket_count > 1"""
execute_query(q)
```

```python
# 18. Display the total number of support tickets available in the system.
q = """select count(*) as total_tickets from support_tickets"""
execute_query(q)
```

```python
# 19. Display the number of tickets for each status.
q = """select status, count(*) as ticket_count from support_tickets group by status"""
execute_query(q)
```

```python
# 20. Display the number of tickets for each priority level.
q = """select priority, count(*) as ticket_count from support_tickets group by priority"""
execute_query(q)
```

```python
# 21. The IT manager wants a report showing each department and the number of support tickets raised by that department. Retrieve this information from the database.
q = """select e.department, count(s.ticket_id) as ticket_count from employees e join support_tickets s on e.employee_id = s.employee_id group by e.department"""
execute_query(q)
```
