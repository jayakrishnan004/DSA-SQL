```python
import pandas as pd;
import mysql.connector as mysql;
```

```python
# Set up the connection details and initialize the cursor for db connection.
name = "root"
password = "jkmysql@123#"
host = "localhost"
port = 3306
database = "students_schema"

connection = mysql.connect(user=name, password=password, host=host, port=port, database=database)
db_info = connection.get_server_info()
print("MySQL Server Information:", db_info)

cursor = connection.cursor(buffered=True)
```

```python
# Execute the sql query.

cursor.execute("select * from students;")
rows = cursor.fetchall()

# Display the results.
for row in rows:
    print(row)
```

```python
# column_names = ["id", "name", "sem", "dept", "loc"];
# df = pd.DataFrame(rows, columns=column_names);
# print(df);

###### OR ######

df = pd.DataFrame(rows, columns = cursor.column_names)
print(df);
```

```python
# To see all databases.
cursor.execute("select * from information_schema.schemata;")
cursor.fetchall()
```

```python
# To list all the columns of table students under students_schema database.
cursor.execute("select column_name from information_schema.columns where table_name = 'students' and table_schema = 'students_schema';")
cursor.fetchall()
```

```python
# Another way to retrieve the columns.
cursor.execute("select * from students_schema.students limit 0;")
columns = [column[0] for column in cursor.description]
print(columns)
```

```python
# Once the work is done, close the cursor and connection.
cursor.close()
connection.close()
```
